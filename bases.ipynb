```python
pqrs['fecha'] = pd.to_datetime(pqrs['PERIODO'].astype(str), format='%y%m')
```

```python
pqrs['mes'] = pqrs['fecha'].dt.to_period('M')
```

```python
pqrs_mensual = pqrs.groupby(['mes', 'LOCALIDAD'], as_index=False)['CANTIDAD'].sum()
pqrs_mensual
```

```python
pqrs_mensual_2025 = pqrs_mensual[pqrs_mensual['mes'].dt.year == 2025]
pqrs_mensual_2025
```

```python
pqrs_localid = pqrs_mensual_2025.groupby('LOCALIDAD', as_index=False).agg(num_pqrs=('CANTIDAD', 'sum'))
pqrs_localid.sort_values(by='num_pqrs', ascending=False)
```
